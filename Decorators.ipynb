### Decorators:
A decorator in Python is a design pattern that allows you to modify or extend the behavior of a function or method without changing its actual source code

```python
# function copy
# closures
# decorators
```

#### function copy

```python
# function copy

def welcome():
  return 'Hello welcome to our learning app'

wel = welcome # copy this function inside a varible
wel()

del welcome
wel()
```

#### closures: 
function inside function

```python
''' what we implement or create inside a main method 
that can be accessed by sub_method  '''

def main_welcome():
  msg = 'Welcome'
  print('This show is fire')
  def sub_welcome():
    print('welcome to show')

    print(msg + f' to our learning app')
  return sub_welcome()
```

```python
main_welcome()

```

#### function that takes function as an argument

```python
def Main_Welcome(func,x,y):

  def Sub_welcome():
    print('welcome to the advanced course')
    result = func(x,y)
    print(result)
    print('Please learn it properly')
    return result
  return Sub_welcome()
  
    


def add_func(x,y):
  return x+y

def multiply(x,y):
  return x*y

```

```python
print(Main_Welcome(add_func,4,6))
```

```python
Main_Welcome(multiply,5,5)
```

### decorators

```python

@Main_Welcome
def multiply(x,y):
  return x*y

```

```python
def my_decorator(func):
  def wrapper():
    print('Something is happening before function called')
    func()
    print('Something is happening after function called')
  return wrapper()
```

```python
@my_decorator
def say_hello():
  print('Hello')
```

#### Decorators with arguments

```python
def repeat(n):
  def decorator(func):
    def wrapper(*args,**kwargs):
      for _ in range(n):
        func(*args,**kwargs)

    return wrapper
  return decorator 
```

```python
@repeat(3)
def say_hello():
  print('hello')
```

```python
say_hello()
```
