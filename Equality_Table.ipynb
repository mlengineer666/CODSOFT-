```python
def classify_equality(num):
  abs_number = abs(num)
  if abs_number <= 10:
    return 'Fair'
  elif abs_number <= 20:
    return 'Unfair'
  else:
    return 'Highly Discriminative'

df['Equality Class'] = df['Equality Score'].apply(classify_equality)

df[['Equality Score' , 'Equality Class']].head(10)
```
