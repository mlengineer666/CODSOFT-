Create a programme to change a string element in specified Numpy array to uppercase,lowercase, capitalise the first letter , title-case or swapcase

```python
import numpy as np 
my_arr = np.array(['i','love','kolkata'])
```

```python
type(my_arr)
np.shape(my_arr)
```

```python
np.char.upper(my_arr)   # Upper case
```

```python
np.char.lower(my_arr)  # Lower case
```

```python
np.char.capitalize(my_arr)
```

```python
np.char.title(my_arr) # title method same as capitalize
np.char.swapcase(my_arr)
```
