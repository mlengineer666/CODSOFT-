get row numbers of numpy array if any element is having larger value than a specific value

```python
import numpy as np

arr = np.array([[1,2,3,4],[2,5,7,9],[10,30,406,60],[13,5,63,67]])
np.vstack(arr)
```

```python
given_element = 10
```

```python
np.where(np.any(arr > given_element,axis=0))  # axis 0 is column
```

```python
np.where(np.any(arr > given_element,axis=1))  # axis 1 is row
```
