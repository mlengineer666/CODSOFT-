How to remove rows whic contain null values in numpy array

```python
import numpy as np 
arr = np.array([[np.nan,2,3],[4,6,7],[5,np.nan,0],[48,63,np.nan]])
arr
```

```python
arr.shape
```

```python
nan_value = np.isnan(arr).any(axis=1)
np.where(nan_value)
```

```python
arr_clean = arr[~nan_value]
arr_clean
```
