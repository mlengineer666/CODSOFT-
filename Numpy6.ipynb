How to select minimum value along the rows

```python
import numpy as np 
array = np.array([[[np.nan,2,3,4],[5,3,2,1],[40,13,45,12],[32,11,66,33]],
[[13,24,3,4],[5,31,2,1],[40,13,45,12],[2,11,66,33]],
[[13,22,13,4],[5,13,2,1],[40,13,45,12],[3,11,66,33]],
[[10,2,3,4],[54,3,532,1],[40,13,4,45],[10,22,35,24]]])

array
```

```python
array.shape
```

```python
np.apply_along_axis(lambda x: np.min(x), arr =  array,axis=1)
```
