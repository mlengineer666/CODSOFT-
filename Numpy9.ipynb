create 4x4 array with random values , now create a new array from the said array such that it starts with the last row and goes till the first.

```python
import numpy as np 

arr1 = np.arange(16)
arr1.shape
arr1
```

```python
arr1 = arr1.reshape(4,4)
arr1
```

```python
new_array = arr1[::-1]
```
