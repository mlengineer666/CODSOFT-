### Abstraction:
Abstraction is the concept of hiding complex implementation details and shoing only necessary features of an object. This helps to reduce complexity and effort

when we apply abstractmethod or abstractmethod decorator that should be empty,
but those class will inherit the abstractmethod class that class should define that specific method which present in abstractmethod class 

if we don't define the abstract class method we get error, so when to inherit the abstract class make sure that we must have to define that method

```python
from abc import ABC,abstractmethod

## Abstract Base class
class Vehicle(ABC):
  def drive(self):
    print("The vehicle is used for driving")

  @abstractmethod
  def start_engine(self):
    pass

class Car(Vehicle):
  def start_engine(self):
    print('Car Engine started')
  
  

def operate_vehicle(vehicle):
  vehicle.start_engine()
  vehicle.drive()    


car = Car()
operate_vehicle(car)
```
