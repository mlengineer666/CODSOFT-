Encapsulation and Abstraction are two fundamental principle of OOP that help in dsigning robust,maintainable and reusable code. 

### Encapsulation
Encapsulation is the concept of rapping data and methods together as a single unit. it restrict direct access to some of the Objects's component, which is a means of preventing accidental inferences and misuse of the data


Encapsulation with getter and setter method
##### public , protected and private varible these are the access method

```python
class Person:
  def __init__(self,name,age):
    self.name = name  # Public Variable
    self.age = age    # Public Variable
  

def get_name(person):
    return person.name

person = Person("Prabir",55)
get_name(person)
```

for accessing without knowing the actual private varible we have getter and setter

```python
class Person:
  def __init__(self,name,age):
    self._name = name   # protected variable
    self._age = age   # protected variable

person = Person('Prabir',55)
dir(person)
person._name
```

```python
''' Encapsulation with getter and setter '''
''' we have to hide the private attributes and that's we use getter and setter '''
class Person:
  def __init__(self,name,age):
    self.__name = name    # private attribute
    self.__age = age    # private attribute
  
  def get_name(self):   # getter for public
    return self.__name
  
  def set_name(self,name):  # set the new name
    self.__name = name 

person = Person('Sagarika',40)
person.get_name()
person.set_name('Prabir')
person.get_name()



```
