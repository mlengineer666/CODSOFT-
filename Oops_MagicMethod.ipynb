Magic methods (also called dunder methods, short for double underscore) are special, built-in methods in Python that start and end with two underscores, such as __init__ or __str__

Magic method is used to define the behaviour of objects for buit-in operations such as arithmetic , comparision and more

```python
# basics magic method
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age


person = Person('Prabir',20)   
print(person)
```

```python
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age
  
  def __str__(self):
    return f"{self.name}, {self.age} years old"


  

person = Person('Prabir',20)   
print(person)  # it can now print the output , previously it prints the address


```

### Object Overloading

```python
'''
__add__(self, other)  Addition
__sub__(self, other)  Subtraction
__mul__(self, other)  Multiplication
__eq__(self, other)   Equality comparison
__lt__(self, other)Less than comparison
__str__(self)Readable string representation

'''
```

```python
# Mathematical Operations on vector
class Vector:
  def __init__(self,x,y):
    self.x = x
    self.y = y
  
  def __add__(self,others):
    return Vector(self.x+others.x , self.y+others.y)

  def __mul__(self,others):
    return Vector(self.x*others.x , self.y*others.y)
  
  def __sub__(self,others):
    return Vector(self.x-others.x , self.y-others.y)
  
  def __eql__(self,others):
    return self.x == others.x, self.y == others.y
  
  def __repr__(self):
    return f'Vector({self.x}, {self.y})'


v1 = Vector(4,5)
v2 = Vector(6,7) # other 

v1+v2
v1-v2
```

```python
# For ComplexNumber overloading
class ComplexNumber:
  def __init__(self,real,img):
    self.real = real
    self.img = img

  def __add__(self,other):
    return ComplexNumber(self.real+other.real, self.img+other.img)
  
  def __truediv__(self,other):
    denominator = self.real**2 + other.img**2
    real_part = (self.real * other.real + self.img * other.img) / denominator
    img_part = (self.real * other.real - self.img * other.img) / denominator
    return ComplexNumber(real_part , img_part)
  
  def __repr__(self):
    return f'{self.real} + {self.img}i'

complexnum1 = ComplexNumber(3,6)
complexnum2 = ComplexNumber(4,7)

complexnum1 + complexnum2
complexnum1 / complexnum2

```
