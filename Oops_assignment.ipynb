#### Lesson: Single and Multiple Inheritance

```python
''' Assignment 1: Single Inheritance Basic
Create a base class named Animal with attributes name and species. Create a derived class named Dog that inherits from Animal and adds an attribute breed. Create an object of the Dog class and print its attributes.

 '''

class Animal:
  def __init__(self,name,species):
    self.name = name
    self.species = species

class Dog(Animal):
  def __init__(self,name,species,breed):
    super().__init__(name,species)
    self.breed = breed

def print_attributes(dog):
  print(dog.name, dog.species , dog.breed)

  

dog = Dog('jimmy','Canine', 'Golden-Retriever')
print_attributes(dog)

```

```python
''' Assignment 2: Method Overriding in Single Inheritance
In the Dog class, override the __str__ method to return a string representation of the object.
Create an object of the class and print it. '''


class Animal:
  def __init__(self,name,species):
    self.name = name
    self.species = species
  
  
  
class Dog(Animal):
  def __init__(self,name,species,breed):
    super().__init__(name,species)
    self.breed = breed

  def __str__(self):
    return f'Name: {self.name}, dog_species: {self.species}, dog_breed : {self.breed}'
  
dog = Dog('Buddy', 'Canine', 'Golden Retriever')

print(dog)

```

```python
''' 
Assignment 3: Single Inheritance with Additional Methods In the `Dog` class, 
add a method named `bark` that prints a barking sound. 
Create an object of the class and call the method.
'''

class Dog(Animal):
  def __init__(self,name,species,breed):
    super().__init__(name,species)
    self.breed = breed

  def __str__(self):
    return f'Name: {self.name}, dog_species: {self.species}, dog_breed : {self.breed}'
  
  def bark(self):
    print('woof woof!')


dog = Dog('kalua','cannie','speach')

dog.bark()
```

```python
### Assignment 4: Multiple Inheritance Basic

''' Create a base class named `Walker` with a method `walk` that prints a walking message.
Create another base class named `Runner` with a method `run` that prints a running message. 
Create a derived class named `Athlete` that inherits from both `Walker` and `Runner`. 
Create an object of the `Athlete` class and call both methods.
'''

# Base class 1
class Walker:
  def walk(self):
    print('Atheletes are walking!')

# Base class 2
class Runner:
  def run(self):
    print('Atheletes are running now!')

# Derived class
class Athelete(Walker,Runner):
  pass

athelete = Athelete()
athelete.walk()
athelete.run()
```

```python
### Assignment 5: Method Resolution Order (MRO) in Multiple Inheritance

''' 
In the `Athlete` class, override the `walk` method to print a different message. 
Create an object of the class and call the `walk` method. 
Use the `super()` function to call the `walk` method of the `Walker` class.
'''

class Walker:
  def walk(self):
    print('Walking!!')

class Runner:
  def run(self):
    print('Running!')
  
class Athelete(Walker,Runner):
  def walk(self):
    print('Athelete class also derived the walk method')
    super().walk()  
  
athelete = Athelete()
athelete.walk()



```

```python
### Assignment 6: Multiple Inheritance with Additional Attributes

''' In the `Athlete` class, add an attribute `training_hours` and a method `train` that prints the training hours. 
Create an object of the class and call the method.
'''

class Athelete(Walker,Runner):

  def __init__(self,training_hours):
    self.training_hours = training_hours

  def walk(self):
    print('Athelete class also derived the walk method')
    super().walk()  
  
  def train(self):
    print(f'Today progress in training {self.training_hours} hours!')

athelete = Athelete(6)

athelete.train()
```
