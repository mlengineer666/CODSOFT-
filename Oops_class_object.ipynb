Class with Constructor

Create a class named Student with attributes name and age. Use a constructor to initialize these attributes. Create an object of the class and print its attributes.

```python
class Student:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def print_attribute(self):
    print(f'Name of the student is {self.name} and age is {self.age}')


student = Student('Sandipan',21)
student.name
student.age
student.print_attribute()
```

Class with Private Attributes


```python
class BankAccount:
  def __init__(self,account_number,balance):
    self.__account_number = account_number
    self.__balance = balance

  def deposit(self,acc_number,amount):
    if self.__account_number == acc_number:
      self.__balance += amount
    else:
      print("Invalid Account Number")
  
  def withdraw(self, acc_no , amount):
    if self.__account_number == acc_no and amount > self.__balance:
      print("Insufficient balance")
    elif self.__account_number != acc_no:
      print('Invalid Account Number')
    else:
      self.__balance -= amount

      
  
  def show_balance(self):
    print(f'Current balance {self.__balance}')

  

bank = BankAccount(25300122059,1000000)
bank.show_balance()
bank.deposit(203004040,3938)
bank.show_balance()

bank.withdraw(25300122059,22222)
bank.show_balance()


bank._BankAccount__account_number  # private attribute access
```
