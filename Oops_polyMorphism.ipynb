PolyMorphism:

Polymorphism is a core concept in oop that allows objects of different classes to be treated as objects of a common superclass.typically achieved by method overriding and interfaces.

Method Overriding:

Method overriding allows a child class to provide a specific implementation of a method that is already defined in parent class.



```python
''' Base class '''
class Animal:
  def speak(self):
    return f'Sound of the Animal'

''' Derived class1 '''
class Dog(Animal):
  def speak(self):
    return 'woof!'

''' Derived class1 '''
class Cat(Animal):
  def speak(self):
    return 'Meow!'


dog = Dog()
cat = Cat()
print(dog.speak())
print(cat.speak())
```

PolyMorphism with Functions and Methods

```python
''' Base Class '''
class Shape:
  def area(self):
    return "The areas of the figure"


''' Derived Class1 '''
class Rectangle(Shape):
  def __init__(self,width,height):
    self.width = width
    self.height = height

  def area(self):
    return self.width * self.height


''' Derived Class2 '''
class Circle(Shape):
  def __init__(self,radius):
    self.radius = radius

  def area(self):
    return 3.14 * self.radius ** 2



''' Function that demonstrate Polymorphism and it is outside of all classes '''
def print_area(shape):
  print(f'The area is {shape.area()}')



rectangle = Rectangle(4,6)
print_area(rectangle)

circle = Circle(5)
print_area(circle)
  
```

#### Interfaces is call in Abstract base class

Abstract base class (ABC) are used to define common methods for a group of releated objects,they enforce derived classes to implement particular methods , promoting consitency accross different implementations.

```python
''' we need to import abstract method and abstractclassmethod is decorator '''
from abc import ABC , abstractclassmethod

# Define Abstract class
class Vehicle(ABC):

  @abstractclassmethod
  def start_engine(self):
    ...


# Derived class1
class Car(Vehicle):
  def start_engine(self):
    print(f'Car engine started')

# Derived class2
class Motorcycle(Vehicle):
  def start_engine(self):
    print(f'Mototrcycle engine started')


car = Car()
motorcycle = Motorcycle()

car.start_engine()
motorcycle.start_engine()

```
