```python

matrix = np.array(np.arange(1, 10)).reshape(3,3)

for row in matrix:
    print(row)
  
print(f'Second row, third column element is :... {matrix[1][2]}')
```

```python
# Transpose of the matrix
def transpose_matrix(matrix):
  transpose = [[matrix[j][i] for j in range(len(matrix))] for i in range (len(matrix[0])) ]
  return transpose

transposed_mat = transpose_matrix(matrix)
print('Original matrix:\n', matrix)
print('Transposed matrix:\n', transposed_mat)
for transposed_row in transposed_mat:
    print(transposed_row)
```

```python
def flatten_list(nested_list):
  flat_list = [item for sublist in nested_list for item in sublist]
  return flat_list

nested_list = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]
flattened_list = flatten_list(nested_list)
print('Nested list:', nested_list)

print('Flattened list:', flattened_list)

```

```python
l1 = [1,2,3,4,5]
l2 = [9,8,7,6,5]

combine_list = list(zip(l1,l2))
list(combine_list)
```

```python
def reverse_list(lst)->list:
  return lst[::-1]

print(f'reversed list {reverse_list(l1)}')
print(f'original list {l1}')
```

```python
def rotate_list(lst, n):
  n = n % len(lst)  # Handle cases where n is greater than the list length
  return lst[-n:] + lst[:-n]

print(f'rotated list {rotate_list(l1, 3)}')
```

```python
def list_intersection(lst1,lst2):
  return lst1 & lst2

print(f'Intersection of l1 and l2 is {list_intersection(set(l1),set(l2))}')
print(f'original list l1 is {l1}')
print(f'original list l2 is {l2}')
```
