```python
''' 
Tuples are immutable and lists are mutable.
Tuples are defined using parentheses () and lists are defined using square brackets [].
Tuples can be used as keys in dictionaries, while lists cannot.
Tuples have a fixed size, while lists can grow and shrink in size.  
Tuples are generally faster than lists for certain operations, such as iteration and access, due to their immutability.
Lists are more versatile and can be modified, while tuples are more suitable for representing fixed collections of items. 
In summary, tuples are immutable and have a fixed size, while lists are mutable and can grow or shrink in size.
The choice between them depends on the specific use case and whether immutability is desired.
  '''

''' once tuple is created it cannot be modified.'''

lst1 = [1,2,3,4,5]
lst1[2] = 'Sandi'
print(lst1)

my_tuple = (1,2,3,4,5)
try:
  my_tuple[2] = 'sandi' # This will raise a TypeError because tuples are immutable and cannot be modified after creation.
except:
  pass

# tuple methods
print(my_tuple.count(14))
print(my_tuple.index(3))


# Pacing and unpacking tuples
packed_tuple = '1', 'sandi',123,3.14
print(packed_tuple)

a,b,c,d = packed_tuple
print(f'{a}')
print(f'{b}')
print(f'{c}')
print(f'{d}')

# *args
my_tuple = (1,2,3,4,5)
a , *args, b = my_tuple
print(f'{a}')
print(f'{args}')
print(f'{b}')

# **kwargs
def my_function(**kwargs):
  for key, value in kwargs.items():
    print(f'{key}: {value}')  

my_function(name='sandi', age=30, city='New York')

# nested tuples
nested_tuple = (1, 2, (3, 4), 5)
nested_tuple[2][0] # Accessing the first element of the nested tuple (3, 4) which is 3

for i in nested_tuple:
    print(i)
```

```python
def min_in_tuple(tpl):
    return min(tpl)

def max_in_tuple(tpl):
    return max(tpl)

def sum_of_tuple(tpl):
    return sum(tpl)

sample_tpl = (1, 2, 3, 4, 5)
print(f"Minimum: {min_in_tuple(sample_tpl)}")
print(f"Maximum: {max_in_tuple(sample_tpl)}")
print(f"Sum: {sum_of_tuple(sample_tpl)}")
```
