```python
# sets membership test
sets = {1,2,4,5}
print(3 in sets)
print(1 in sets)
```

```python
# Mathematical Operation
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,1,9,10}

# Union
set1.union(set2)

# Intersection
set1.intersection(set2)

print(set1)
print(set2)
# Intersection_update
set1.intersection_update(set2) # those which are not in set1 but in set2 that's 


```

```python
# Difference
set1 = {1,2,4,47,6}
set2 = {1,2,3,4,5}
set1.difference(set2)
```

```python
# Symmetric Difference
set1.symmetric_difference(set2) # which elements are common
```

```python
print(set1.issubset(set2))
print(set2.issubset(set1))

print(set1.issuperset(set2))
```
