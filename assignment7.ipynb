Dictionary

Dictionary are unordered collection of items. store data in key-value pairs, key must be unique and immutable(e.g- strings,number or tuples), while values can be of any type

```python
my_dict = {}
print(type(my_dict))
```

if duplicates key is present then the next value of that key will be assigned

```python
my_dict = {'name':'sandipan','age':21,'name':43,'name':'sandip'}
print(my_dict)
```

```python
my_dict = {'name':'sandipan','age':21,'year':3,'year':31}  
print(my_dict)
```

```python
my_dict = {'name':'sandipan','age':21,'year':3,'grade':'A+'}
print(my_dict)

my_dict['name']
my_dict['age']

#  access element by get method -> pass the KEY
my_dict.get('name')
my_dict.get('age')

my_dict.get('last-name','not-available')
```

dictionary is mutable

```python

my_dict['age'] = 9
my_dict

my_dict['address'] = 'India'
my_dict
```

```python
try:
  del my_dict['year']
except KeyError :
  my_dict
# raise KeyError
```

```python
dict_key = (my_dict.keys())
dict_value = (my_dict.values())
```

```python
print(dict(zip(dict_key,dict_value)))
```

```python
# all key-value pairs
my_dict.items()
```

```python
my_dict_copy = my_dict # it create permanant copy
print(my_dict)
print(my_dict_copy)

# shallow copy
my_dict_copy = my_dict.copy() # it will be unchanged if value are changes inside the copy-dictionary 
```

```python
my_dict.update({'name':'palan'})
my_dict
```

```python
# update value by merge operator ( |= )
val = {'role':'Manager','department':'HR'}
my_dict_copy |= val

my_dict_copy
```

```python
# iterating
my_dict

# iterate over keys
for keys in my_dict.keys():
  print(keys)

print('+++++++++++++++++++++++++++++++++++++++')

# iterate over vales
for val in my_dict.values():
  print(val)
print('+++++++++++++++++++++++++++++++++++++++')

# iterate over keys.values
for key,val in my_dict.items():
  print(key, val)
```

```python
# Dictionary Comprehension
squares = {x:x**2 for x in range(5)}  # here x is key and x**2 value
print(squares)
```

```python
squares_even = {x:x**2%2==0 for x in range(5)}  # Method1
print(squares_even)


squares_even = {x:x**2 for x in range(5) if x%2==0}  # Method2
print(squares_even)
```

```python
numbers = [1,2,2,2,3,3,3,3,3,4,4,4]
frequency = {}
for x in numbers:
  if x in frequency:
    frequency[x]+=1
  else:
    frequency[x] = 1

print(frequency)

```

```python
# Merge Dictionary with the help of keyword-arguments

dict1 = {'a':1,'b':2}
dict2 = {'b':5,'c':3}

merged_dict = {**dict1 , **dict2}
merged_dict
```
