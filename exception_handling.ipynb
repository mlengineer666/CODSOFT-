try,except and else block, after try block successfully execute else part will be definitely executed but when except or error happen else part won't execute



```python
try:
  num = int(input("Enter the number"))
  res = 10/num
except ValueError:
  print("That's not a valid number")
except ZeroDivisionError:
  print("You can't divide by zero!")
except Exception as ex:
  print(ex)
else:
  print(f"result is {res}")
```

try,except,else and finally. finally block excute every time whether it is error or correct

```python
try:
  num = int(input("Enter the number"))
  res = 10/num
except ValueError:
  print("That's not a valid number")
except ZeroDivisionError:
  print("You can't divide by zero!")
except Exception as ex:
  print(ex)
else:
  print(f"result is {res}")
finally:
  print("All are good")
```
