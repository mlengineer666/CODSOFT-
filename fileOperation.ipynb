```python
def count_text_file(file_path):
  with open(file_path,'r') as file:
    lines = file.readlines()
    line_count = len(lines)
    word_count = sum(len(line.split()) for line in lines)
    char_count = sum(len(char) for char in lines)
  return f'lines: {line_count}, words: {word_count}, characters: {char_count}'


count_text_file('sample.txt')
```

```python
def read_file(file):
  with open(file,'r') as f:
    for line in f:
      print(line.strip())

read_file('sample.txt')
```

```python
def write_file(lines , file):
  with open (file,'w') as f:
    for line in lines:
      f.write(line)

line =[' Hey i am writing code in vsCode',
'learning about files,'
'then after maybe other topic']

write_file(line,'output.txt')
```

```python
with open('output.txt','r') as f:
  for line in f:
    print(line)
```

```python
def copy_file(source_file,destination_file):
  with open(source_file,'r') as s_file:
    with open(destination_file,'w') as d_file:
      d_file.write(s_file.read())

copy_file('output.txt','new_file')     
```

```python
with open('new_file','r') as f:
  for line in f:
    print(line.strip())
```

```python
def appending(line,log_file):
  with open(log_file,'a') as file:
    file.write(line)

appending('\nThis is new log entry','new_file')
```

```python
with open('new_file','r') as f:
  for line in f:
    print('\n'+line.strip())
```

```python
def counting_word(file):

  with open(file,'r')as f:
    text = f.read()
    words = text.split()
    return len(words)

counting_word('sample.txt')
```

```python
def find_and_replace(filename,old_word,new_word):
  
  with open(filename,'r')as f:
    content = f.read()
  new_text = content.replace(old_word,new_word)
  with open(filename,'w') as f:
    f.write(new_text)
  

find_and_replace('source.txt','how','who')
```

```python
def read_reversed(filename):
  with open(filename,'r') as file:
    lines = file.readlines()
  for line in reversed(lines):
    print(line.strip())

read_reversed('sample.txt')
```

```python
def count_lwc(filename):
  with open(filename,'r') as file:
    line_count = file.readlines()

    word_count = sum(len(word.split()) for word in line_count)

    character_count = sum(len(char) for char in line_count)


  return len(line_count),word_count,character_count

count_lwc('sample.txt')
```
