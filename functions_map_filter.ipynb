Filter out elements based on conditions

```python
def even_num(num):
  return True if num%2==0 else False

even_num(6)
```

```python
lst = [1,2,3,4,5,6,7,8,9]
list(filter(even_num,lst))  # it will select those elements which are stisfied this function

# Filter with lambda function
list(filter(lambda x: x > 5, lst))
```

```python
def count_even_odd(lst):
    # Your code goes here
    count_odd = 0
    count_even = 0
    
    even_num = list(filter(lambda x:x%2==0,lst))
    
    odd_num = list(filter(lambda x:x%2!=0,lst))
    
    for i in even_num:
        count_even+=1
    for j in odd_num:
        count_odd+=1
        
    return (count_even,count_odd)

count_even_odd([1,2,3,4,95,-10])
```

```python
def is_subset(lst1, lst2):
  for i in lst1:
    for j in lst2:
      if lst1[i] == lst2[j]:
        return True
      else: 
        return False

lst1 = [1, 2,3]
lst2 = [1, 2, 3, 4, 5]

is_subset(lst1,lst2)
```

```python
def max_consecutive_difference(lst):
    if len(lst) < 2:
        return 0

    maxx = 0
    for i in range(len(lst) - 1):
        diff = abs(lst[i + 1] - lst[i])
        if diff > maxx:
            maxx = diff

    return maxx

max_consecutive_difference([10, 11, 15, 3])
```

```python
def sort_list(lst):
  for i in range(len(lst)-1):
    for j in range(i+1,len(lst)):
      if lst[i] <= lst[j]:
        lst[i] = lst[i]
        lst[j] = lst[j]
      else:
        lst[i],lst[j] = lst[j],lst[i]
  return lst

def merge_sorted_list(l1,l2):
  l3 = []
  i = 0
  j = 0
  while i < len(l1) and j < len(l2):
    if l1[i] <= l2[j]:
      l3.append(l1[i])
      i+=1
    else:
      l3.append(l2[j])
      j+=1


  while i < len(l1):
      l3.append(l1[i])
      i+=1

  while j < len(l2):
      l3.append(l2[j])
      j+=1
  
  return l3



sort1 = sort_list(lst1)
sort2 = sort_list(lst2)

print(merge_sorted_list(sort1,sort2))



lst1 = [1,2,33,4,5]
lst2 = [2,4,55,6]



```
