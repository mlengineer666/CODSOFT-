Temperature Conversion

```python

def convert_temperature(temp,unit):
  if unit == 'C':
    return temp * 1.8 + 32
  elif unit=='F':
    return round((temp-32)*5/9 , 2)
  else:
    return None

print(convert_temperature(25,'C'))
print(convert_temperature(75,'F'))
print(convert_temperature(80,'F'))


```

Password Strength Checker

```python
def password_strength(Password):
  if len(Password) < 7:
    return False
  if not any(char.isdigit() for char in Password):
    return False
  if not any(char.islower() for char in Password):
    return False
  if not any(char.isupper() for char in Password):
    return False
  if not any(char in '!@#$%^&*()_+~' for char in Password):
    return False
  return True


print(password_strength('As7cii@'))
print(password_strength('weekday@2'))

```

Calculate total cost of items in a Shopping cart



```python
def calculate_total_cost(cart):
  total_cost = 0
  for cost in cart:
    total_cost+=cost['price'] * cost['quantity'] 
  
  return f'{round(total_cost*10,2)} is your total cost of your cart.'



cart = [
  {'name':'apple','price':0.7,'quantity':4},
  {'name':'banana','price':0.5,'quantity':6},
  {'name':'guava','price':0.8,'quantity':3}
]

calculate_total_cost(cart)
```

Check if a String is Palindrome

```python
def check_palindrome(s):

  # s = s.lower().replace(' ','')
  low = 0
  high = len(s) - 1 

  while low < high:
    if s[low] != s[high]:
      return f'{s} is not a palindrome'
    low+=1
    high-=1
  return f'{s} is a palindrome'


print(check_palindrome('jjk'))
print(check_palindrome('hello'))
print(check_palindrome('radar'))
print(check_palindrome('racecar'))
```

Factorial of number using recursion

```python
def factorial_of_number(n):
  if n == 0 :
    return 1
  return n*factorial_of_number(n-1)

n = int(input('Enter the number'))
factorial_of_number(n)
```

A Function to read a file and count the frequency of each word

```python
def count_word_frequncy(file_path):
  word_count = {}
  with open(file_path,'r')as file:
    for line in file:
      words = line.split()
      for word in words:
        word = word.lower().strip('''.,!;:1234567890"\'  /''')
        word_count[word] = word_count.get(word,0) + 1  # get methode in dictionary takes key,initial_value as an argument

  return word_count


file_path = 'sample.txt'

count_word_frequncy(file_path)
```

Validate E-mail

```python
import re # which is regular expression

def valid_email(email):
  pattern = r'^[a-zA-Z0-9_.+-]+@[a-zA-Z]+\.[a-z]+$'
  return re.match(pattern,email) is not None


print(valid_email('sandipan@gmail.com')) # True
print(valid_email('sandipan@gmail.')) # False
print(valid_email('sandipangmail.com')) # False
print(valid_email('sandipan0001gmail@yahoo.com'))   # True
print(valid_email('sandipan@gmail9.com')) # False
print(valid_email('sandipan@gmail.COM'))  # False


```
