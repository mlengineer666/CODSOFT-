generators are simpler way to create iterators. They Use the yeild keyword to produce a series of values lazily , which means they generate value on the fly and do not store them in memory. 

```python
def square(n):
  for i in range(n):
    return i**2

square(3)
```

```python
def square(n):
  for i in range(n):
    yield i**2

squares = square(3)

for i in squares:
  print(i)
```

using yield multiple value can be returned

```python
def my_generator():
  yield 1
  yield 2
  yield 3

gen = my_generator()
gen
```

```python
next(gen)
```

```python
# Practical example:
# Reading large files

def read_large_file(file_path):
  with open(file_path,'r') as file:
    for line in file:
      yield line


large_file = 'D:\ML\DATA_SCIENCE\large_file.text'
```

```python
for line in read_large_file(large_file):
  print(line.strip())
```
