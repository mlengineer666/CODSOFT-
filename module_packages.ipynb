```python
# Data serialization
'''  json.dumps() converts a Python object into a JSON string, 
    while json.loads() converts a JSON string back into a Python object.
'''


import json as j
data = {'name':'sandipan','age':21,'gender':'Male'}

json_str = j.dumps(data)
print(json_str)
print(type(json_str))

parsed_data = j.loads(json_str)
print(parsed_data)
print(type(parsed_data))

```

```python
# csv
import csv
with open('example.csv',mode='w',newline='') as file:
  writer = csv.writer(file)
  writer.writerow(['name','age','gender'])
  writer.writerow(['sandipan',21,'Male'])

with open('example.csv',mode='r') as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
```

```python
# datetime
''' timedelta use to calculate previous date as well as next date  '''

from datetime import datetime , timedelta

now = datetime.now()
print('Current date ->',now)

yesturday = now - timedelta(days=1)
print('yesturday date ->',yesturday)

tomorrow = now + timedelta(days=1)
print('Tomorrow date ->', tomorrow)

```

```python
import re

pattern = r'\d+'   # d represent the digits and d+ check that any furtherdigit is present or not but it check continuous 
text = 'There are 123 apples 456'

matched = re.search(pattern,text)
matched.group()
```
